```python
#Write a function fibseries (n) that returns a list of all fibinacci series up to n.
list=[0]
x=0
y=1
n=int(input("Enter a number: "))

def fibseries(n):
    global x,y
    for i in range(n):
        next_num=x+y
        x=y
        y=next_num
        list.append(x)
    print(list)
fibseries(n)
```

```python
#Write a function merge_dict(dict1, dict2) that merges
#two dictionaries dict1 and dict2.
#If they have overlapping keys, the values from dict2
#should overwrite those from dict1.

def merge_dict(dict1,dict2):
    merged_dict = dict1.copy()
    merged_dict.update(dict2)
    return merged_dict
dict1 = {'a': 1, 'b' : 2, 'c':3}
dict2={'b':3, 'd':4}
result = merge_dict(dict1,dict2)
print(result)
```

```python
#Write a function generate_permutations(s) that generates all
#permutations of a string s and returns
#them as a list.

def generate_permutations(s):
    if len(s) == 0:
        return ['']
    perms = []
    for i in range(len(s)):
        for p in generate_permutations(s[:i] + s[i+1:]):
            perms.append(s[i] + p)
    return perms

s = "hey"
print(generate_permutations(s))
```

```python
#Write a function custom_sort(lst) that sorts a list of strings based
#on their lengths, using a custom key function in sorted().

def custom_sort(list_):
    return sorted(list_,key=len)
list=["black","blue","red","green","yellow","violet","white","indigo"]
sorted_list=custom_sort(list)
print(sorted_list)
```
