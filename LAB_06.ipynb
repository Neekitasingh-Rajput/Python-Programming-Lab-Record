```python
1. Create a dictionary from user input.

2. Rename a key of a dictionary.

3. Write a Python program to extract a dictionary with each key having a non-numeric value from a given dictionary.
```

```python
#Question 1
#Create a dictionary from user input.

x = {}
n = int(input("How many entry you want: "))
for i in range(n):
    key = int(input("Enter the key: "))
    value = input("Enter the value: ")
    x[key] = value
print("The constructed dictionary is: ",x)
```

```python
#Question 2
#Rename key of a dictionary.

Dict1 = {"name":"neekita"}
Dict1["Username"] = Dict1.pop("name")
print(Dict1)
```

```python
#Question 3
#Python program to extract dictionary with each key having non-numeric value from a given dictionary.

Dict2 = {"name" : "Neekita" , "class" : "BscDM", "sem" : 1}

for i in Dict2:
    value = Dict2[i]
    if str(value).isnumeric():
        print({i: Dict2[i]})
```
