```python
1. Create a list and perform the following methods:

    1) insert()
    2) remove()
    3) append()
    4) pop()
    5) clear()

2. Create a list from user input and sort the elements in any order.

3. Write a Python program to find the length of a list.

4. Write a Python program to find the smallest and largest number in the list.
```

```python
#Ques.1

a = [11,12,13,14]
a.insert(14,15)#will add the given value to the list after that value.
print(a)
a.remove(11)#will remove the value from list.
print(a)
a.append(16)#will add this value to the list.
print(a)
a.pop(0)#will remove the value of that position.
print(a)
a.clear()#will clear the entire list.
print(a)
```

```python
#Ques.2
length = int(input("Enter the length of your list: "))
b = []
for i in range(length):
    num = int(input(f"Enter number {i + 1}: "))
    b.append(num)

print("Original list:", b)

# Sort the list in ascending order
b.sort()
print("List sorted in ascending order:", b)

# Sort the list in descending order
b.sort(reverse=True)
print("List sorted in descending order:", b)
```

```python
#Ques.3
c = ["a", "b", "c", "d"]
print("The length of the list is" ,len(c))
```

```python
#Ques.4
d = [11,22,33,44,55]
largest = smallest = d[0]
for i in d:
    if i > largest:
        largest = i
    elif smallest > i:
        smallest = i
print("The largest no of the list is", largest)
print("The smallest no of the list is", smallest)
```
