```python
#QUESTION 1
import numpy as np

list_1 = []
n = int(input("Enter the no of lists:"))
for i in range(n):
    list_i = []
    a = int(input("Enter no of elements:"))
    for k in range(a):
        b = int(input("Enter the element"))
        list_i.append(b)
    list_1.append(list_i)
print("List of Lists:", list_1)
print(np.array(list_1))
```

```python
#QUESTION 2

import numpy as np

m = int(input("Enter the number of rows (m): "))
n = int(input("Enter the number of columns (n): "))
k = int(input("Enter the upper bound for random values (k): "))

user_array = np.random.randint(1, k+1, size=(m, n))

print("Random m x n NumPy Array with values between 1 and k:")
print(user_array)
```
