```python
#LAB 1

'''
1. Print a statement, “Hello Christites”

2. Perform basic arithmetic operation on two or more numbers.

3. Solve the following Expressions-
b) 10x2 + 5x + 2xy + y

c) The number of Poetry books on Sraddha’s bookshelf is 5 less than twice the number of Novels. 
Call the number of novels n=any value of your choice, and find the number of poetry books. 
Print Proper message.

d) Calculate the value of C from the following equation, when the value of F is given.
[C/5 = (F-32)/9]
'''



# 1. Print statement
print("Hello Christites")

# 2. Arithmetic operations
a, b, c = 10, 5, 2
print("Addition:", a + b + c)
print("Multiplication:", a * b * c)
print("Division:", a / b)
print("Subtraction:", a - b)

# 3b. Solve expression: 10x² + 5x + 2xy + y
x, y = 3, 2
result = 10 * (x ** 2) + 5 * x + 2 * x * y + y
print("Result of the expression (10x² + 5x + 2xy + y):", result)

# 3c. Poetry books calculation
n = 8
poetry_books = 2 * n - 5
print(f"The number of novels is {n}, and the number of poetry books is {poetry_books}.")

# 3d. Solve for C in the equation
F = 77
C = 5 * (F - 32) / 9
print(f"For F = {F}, the value of C is {C:.2f}.")
```
