```python
#Question 1
#Generate a list of numbers that are palindromes from 1 to 100 using comprehension.

Palindromes = [i for i in range(1,101) if str(i) == str(i)[::-1]]
print("Palindromes in (1,100) are" , Palindromes)
```

```python
#Question 2
#Generate a list of tuples containing a number and its square using Comprehension.

num_squares = [(x , x**2) for x in range(1,6)]
print(" The required list: " , num_squares)
```

```python
#Question 3
#Create a dictionary of words and their lengths from a sentence using comprehension.

sentence = input("Enter a sentence: ")
k = sentence.split()
Dict = {i : len(i) for i in k}
print(Dict)
```

```python
#Question 4
'''
Distinct words and their length, excluding those with lengths not in a Fibonacci series [if the
length of the word is NOT in the series 0 1 1 2 3 5 8 13….], in a sentence using dictionary
comprehension.
[“She is Good Guy”- here len(she)=3, len(is)=2, len(good)=4, and len(Guy)=3. So she, is, Guy
is in Fibonacci series and Good is not. Therefore the output should be {she:3, is:2, Guy:3}]
'''

fib = [0,1]
while fib[-1]<30:
    fib.append(fib[-1] + fib[-2])

s = input("Enter the sentence:")
x = {w:len(w) for w in s.split() if len(w) in fib}
print(x)
```
