```python
1. Write a program to ask the user for their full name. It should give the following output:
    a. The whole name
    b. the first character
    c. the first 5 characters
    d. the fourth, fifth and sixth characters

2. Check a given string is palindrome or not.

3. Write a program which asks two string and check they are 
    1. Equal 2. Less than 3. Greater than 
    the first number

4. Check a particular letter (given by User) is there in a string or not. If it is there, count the number
of appearance.
```

```python
#Question 1
N= str(input("Enter your name: "))
print(N)
print(N[0])
print(N[0:5])
print(N[3:6])
```

```python
#Question 2
s= str(input("Enter a string: "))
if s==(s[::-1]):
    print("The string is palindrome!")
else:
    print("The string is not a palindrome!")
```

```python
#Question 3
a= str(input("Enter a string: "))
b= str(input("Enter another string: "))
if a==b:
    print("The strings are equal!")
elif a>b:
    print("The first string is greater than the second!")
else:
    print("The first string is less than the second!")
```

```python
#Question 4
a= str(input("Enter a string: "))
b= str(input("Enter an element: "))

if b in a:
    print("The element is there in the given string!")
    count = 0
    for i in a:
        if i == b:
            count = count+1
    print("count =" , count)
else:
    print("The element is not there in the given string!")
```
