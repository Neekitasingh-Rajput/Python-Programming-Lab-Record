```python
1. Create a tuple from user input and perform the following methods:

    1) insert()
    2) remove()
    3) append()
    4) pop()
    5) clear()

2. Create a set from user input and perform the following operations:

    1) Union
    2) Intersection
    3) Set Difference
    4) Product
```

```python
#Question 1

n= int(input("Enter the length of your list:" ))
list_1 = []
for i in range(n):
    a= input("Enter an element:")
    list_1.append(a)
    #this will ask user for list elemnts and append it to the list.
    
print(list_1)
list_1.insert(5,6)
print(list_1)
list_1.remove(6)
print(list_1)
list_1.append(7)
print(list_1)
list_1.pop(0)
print(list_1)

print(type(list_1))
list_1= tuple(list_1)
print(list_1)
print(type(list_1))
```

```python
#Question 2

m= int(input("Enter the length of your list:" ))
list_one = []
for i in range(m):
    b= input("Enter an element:")
    list_one.append(b)
    #this will ask user for list elemnts and append it to the list.
list_one = set(list_one)
print("Set one:" , list_one)

l= int(input("Enter the length of your list:" ))
list_two = []
for i in range(l):
    c= input("Enter an element:")
    list_two.append(c)
    #this will ask user for list elemnts and append it to the list.
list_two = set(list_two)
print("Set two:" , list_two)

#union
union = list_one | list_two
print("Union of the two sets is:" , union)

#intersection
intersection = list_one & list_two
print("Intersection of the sets is:" , intersection)

#Set difference
Set_difference = list_one - list_two
print("Set difference of the two is:" , Set_difference)
```
